```python
# IMPORT LIBRARIES
import numpy as np
from scipy.optimize import fsolve, root
import cmath
import matplotlib as mp
from pylab import *
```

```python
# DEFINE BASIC CONSTANTS
c=3e8
pi=2*math.asin(1)
epsilon0=8.85e-12
me=9.1e-31
ec=1.6e-19
```

```python
# PHYSICAL INPUT
wavelength = 800e-9
epsAir=1e0
epsSi0=13.64+0.048j
meffSi=0.18
nuSi=(1.1e-15)**-1
```

```python
# SPP BASIC FUNCTIONS

def betaSPP(wavelength, eps1, eps2):
    # calculate the SPP wave number
    omega=2*pi*c/wavelength
    return map(cmath.sqrt, omega**2 / c**2 * eps1 * eps2 / (eps1 + eps2))

def period(betaSPP):
    ReBeta=map(real, betaSPP)
    return 2*pi/ReBeta
```

```python
# OPTICAL FUNCTIONS

def Drude(wavelength, ne, epsilon, nu, meff):
    # Return the value of dielectric function based on simplified Drude model
    omegap2=ne * ec**2 / (me * meff * epsilon0)
    omega=2*pi*c/wavelength
    return epsilon - omegap2/(omega*omega) * 1/(1+1j*nu/omega)
```

```python
# work on tuples or arrays
test=np.array([1e16,1e25,1e28])
epsilonSi=Drude(wavelength, test, epsSi0, nuSi, meffSi)
print "eps1", epsAir
print "eps2", epsilonSi
beta=betaSPP(wavelength, epsAir, epsilonSi)
for i in beta:
    print i.real
period(beta)
```

```python
# Calculate numerically

period(betaSPP(wavelength, epsAir, Drude(wavelength, 1e28, epsSi0, nuSi)))
```

```python
# PLOT

from matplotlib import rc
rc('font',**{'family':'sans-serif','sans-serif':['Helvetica']})
## for Palatino and other serif fonts use:
#rc('font',**{'family':'serif','serif':['Palatino']})
rc('text', usetex=True)

xmin=1e16
xmax=1e28
Nsteps=100
ne = linspace(xmin,xmax,Nsteps)
x=ne
y=period(betaSPP(wavelength, epsAir, Drude(wavelength, ne.transpose, epsSi0, nuSi)))
plot(x, y)

xlabel('N_e (m^{-3})')
ylabel('\Lambda (m)')
title('SPP period')
grid(True)
savefig("test.png")
show()

```
